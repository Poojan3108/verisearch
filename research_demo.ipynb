```python
from agent.graph import build_graph, format_sources
from IPython.display import Markdown, display

graph = build_graph()
```

```python
question = "What are the latest breakthroughs in solid-state EV batteries?"

final = {}
for update in graph.stream({"question": question}, stream_mode="updates"):
    for node, output in update.items():
        final.update(output)
        for line in output.get("log", []):
            print(line)
```

```python
print("Sub-questions:", final["sub_questions"])
print("Sources:", len(final["sources"]), "| Revisions:", final["revisions"], "| Approved:", final["approved"])
```

```python
display(Markdown(final["draft"] + "\n\n## Sources\n" + format_sources(final["sources"])))
```

```python
from agent.graph import build_graph

graph = build_graph()

questions = [
    "How are hospitals using AI for medical diagnosis?",
    "What is the current state of quantum computing?",
]

for q in questions:
    result = graph.invoke({"question": q})
    print(f"\nQuestion: {q}")
    print(f"Sources: {len(result['sources'])} | Revisions: {result['revisions']} | Approved: {result['approved']}")
```
